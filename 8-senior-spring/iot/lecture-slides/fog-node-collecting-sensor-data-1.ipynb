```python
import time
import random

def fog_node_processor(threshold=30.0, batch_size=10):
    """
    Simulates a Fog Node collecting sensor data, averaging it,
    and conditionally sending it to the Cloud.
    """
    buffer = []
    print(f"--- Fog Node Started (Threshold: {threshold}°C) ---")

    while True:
        # 1. Simulate reading data from a local sensor (Edge layer)
        # In a real app, this might be: sensor.read_temperature()
        current_temp = round(random.uniform(20.0, 40.0), 2)
        buffer.append(current_temp)
        print(f"Local Sensor Reading: {current_temp}°C")

        # 2. Check if we have collected enough readings (the 'Batch')
        if len(buffer) >= batch_size:
            average_temp = sum(buffer) / len(buffer)
            print(f"\n[FOG PROCESSING] Batch complete. Average: {average_temp:.2f}°C")

            # 3. Decision Logic: Filter data before sending to Cloud
            if average_temp > threshold:
                send_to_cloud(average_temp)
            else:
                print("[FOG DECISION] Temperature normal. Data discarded to save bandwidth.")

            # 4. Clear the buffer for the next batch
            buffer = []
            print("-" * 40)

        time.sleep(0.5) # Simulate time between readings

def send_to_cloud(data):
    """
    Simulates a high-latency network call to a Cloud API.
    """
    print(f"[@] CLOUD ALERT: Sending average {data:.2f}°C to AWS/Azure Dashboard...")
    # In reality, you'd use: requests.post(cloud_url, json={'avg': data})

# Run the simulation
if __name__ == "__main__":
    try:
        fog_node_processor()
    except KeyboardInterrupt:
        print("\nFog Node shut down.")
```
