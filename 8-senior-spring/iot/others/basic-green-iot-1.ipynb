```python
import time
import random
from datetime import datetime

class SmartSensor:
    """Represents a smart sensor collecting environmental data."""
    def __init__(self, sensor_id, sensor_type, power_source="battery", energy_consumption_rate=0.01):
        self.sensor_id = sensor_id
        self.sensor_type = sensor_type
        self.power_source = power_source
        self.energy_consumption_rate = energy_consumption_rate  # kWh per hour (example)
        self.operational = True
        self.data_log = []

    def measure(self):
        """Simulates a sensor reading."""
        if self.sensor_type == "temperature":
            return round(random.uniform(15, 30), 1)
        elif self.sensor_type == "humidity":
            return round(random.uniform(40, 70), 1)
        elif self.sensor_type == "light_level":
            return random.randint(0, 1000)
        elif self.sensor_type == "soil_moisture":
            return round(random.uniform(20, 60), 1)
        else:
            return None

    def transmit_data(self, data):
        """Simulates transmitting data (minimal energy)."""
        if self.operational:
            timestamp = datetime.now().isoformat()
            self.data_log.append({"timestamp": timestamp, "sensor": self.sensor_type, "value": data})
            print(f"[{timestamp}] Sensor {self.sensor_id} ({self.sensor_type}): Transmitted {data}")
            # Simulate minimal energy consumption for transmission
            self.consume_energy(0.0001)
            return True
        return False

    def consume_energy(self, duration_hours=1):
        """Simulates energy consumption."""
        if self.operational and self.power_source == "battery":
            energy_used = self.energy_consumption_rate * duration_hours
            print(f"Sensor {self.sensor_id}: Consumed {energy_used:.4f} kWh.")
            # In a real scenario, you'd track battery level and potentially go offline
            pass
        elif self.operational and self.power_source == "solar":
            # Solar powered devices ideally have near-zero net energy consumption over time
            print(f"Sensor {self.sensor_id}: Using solar power.")

    def check_power_status(self):
        """Simulates checking power status (simplified)."""
        if self.power_source == "battery":
            # In a real system, you'd monitor actual battery levels
            if random.random() < 0.01:  # Simulate occasional low battery
                self.operational = False
                print(f"Warning: Sensor {self.sensor_id} battery low. Going offline.")
            else:
                self.operational = True
        elif self.power_source == "solar":
            self.operational = True # Assuming sufficient sunlight
        return self.operational

class SmartIrrigationSystem:
    """Represents a smart irrigation system optimizing water usage."""
    def __init__(self, system_id):
        self.system_id = system_id
        self.soil_moisture_sensors = {}
        self.water_valve_open = False

    def add_sensor(self, sensor):
        if sensor.sensor_type == "soil_moisture":
            self.soil_moisture_sensors[sensor.sensor_id] = sensor
            print(f"Irrigation System {self.system_id}: Added soil moisture sensor {sensor.sensor_id}")
        else:
            print(f"Irrigation System {self.system_id}: Cannot add sensor of type {sensor.sensor_type}.")

    def should_irrigate(self, threshold=30):
        """Checks if irrigation is needed based on soil moisture levels."""
        if not self.soil_moisture_sensors:
            return False

        total_moisture = 0
        active_sensors = 0
        for sensor_id, sensor in self.soil_moisture_sensors.items():
            if sensor.check_power_status():
                moisture = sensor.measure()
                if moisture is not None:
                    sensor.transmit_data(moisture)
                    total_moisture += moisture
                    active_sensors += 1

        if active_sensors > 0:
            average_moisture = total_moisture / active_sensors
            return average_moisture < threshold
        return False

    def control_valve(self, open_valve=True):
        """Controls the water valve."""
        if open_valve and not self.water_valve_open:
            print(f"Irrigation System {self.system_id}: Opening water valve.")
            self.water_valve_open = True
        elif not open_valve and self.water_valve_open:
            print(f"Irrigation System {self.system_id}: Closing water valve.")
            self.water_valve_open = False

    def optimize_irrigation(self):
        """Optimizes irrigation based on sensor data."""
        if self.should_irrigate(threshold=40):
            self.control_valve(True)
            # Simulate irrigation duration (can be further optimized based on more data)
            irrigation_duration = random.randint(1, 3)  # Simulate 1-3 hours of watering
            print(f"Irrigation System {self.system_id}: Irrigating for {irrigation_duration} hours.")
            time.sleep(irrigation_duration)
            self.control_valve(False)
        else:
            print(f"Irrigation System {self.system_id}: Soil moisture sufficient. No irrigation needed.")

if __name__ == "__main__":
    # Simulate battery-powered temperature and humidity sensor
    temp_humidity_sensor = SmartSensor(sensor_id="THS001", sensor_type="temperature", power_source="battery", energy_consumption_rate=0.005)
    humidity_sensor_2 = SmartSensor(sensor_id="HS002", sensor_type="humidity", power_source="battery", energy_consumption_rate=0.003)

    # Simulate solar-powered soil moisture sensors
    soil_sensor_1 = SmartSensor(sensor_id="SMS003", sensor_type="soil_moisture", power_source="solar", energy_consumption_rate=0.001)
    soil_sensor_2 = SmartSensor(sensor_id="SMS004", sensor_type="soil_moisture", power_source="solar", energy_consumption_rate=0.001)

    # Create a smart irrigation system
    irrigation_system = SmartIrrigationSystem(system_id="IS001")
    irrigation_system.add_sensor(soil_sensor_1)
    irrigation_system.add_sensor(soil_sensor_2)

    print("\n--- Simulating Smart Sensors ---")
    for _ in range(5):
        if temp_humidity_sensor.check_power_status():
            temp = temp_humidity_sensor.measure()
            if temp is not None:
                temp_humidity_sensor.transmit_data(temp)
        if humidity_sensor_2.check_power_status():
            humidity = humidity_sensor_2.measure()
            if humidity is not None:
                humidity_sensor_2.transmit_data(humidity)
        soil_sensor_1.measure()
        soil_sensor_2.measure()
        time.sleep(1)

    print("\n--- Simulating Smart Irrigation System ---")
    for _ in range(3):
        print(f"\n--- Time: {datetime.now().strftime('%H:%M:%S')} ---")
        irrigation_system.optimize_irrigation()
        time.sleep(5)
```
