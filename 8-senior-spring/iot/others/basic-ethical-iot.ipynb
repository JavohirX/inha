```python
import time
import datetime

class SmartDevice:
    """Represents a generic smart device with basic ethical considerations."""

    def __init__(self, device_id, device_type, user_consent=False, data_collected=None):
        """
        Initializes a SmartDevice.

        Args:
            device_id (str): Unique identifier for the device.
            device_type (str): Type of the device (e.g., sensor, actuator).
            user_consent (bool): Indicates if the user has consented to data collection.
            data_collected (dict): Dictionary to store collected data (initially empty).
        """
        self.device_id = device_id
        self.device_type = device_type
        self.user_consent = user_consent
        self.data_collected = data_collected if data_collected is not None else {}
        self.last_updated = None

    def request_consent(self):
        """Simulates requesting user consent for data collection."""
        print(f"Device {self.device_id} ({self.device_type}): Requesting your consent to collect data.")
        response = input("Do you consent? (yes/no): ").lower()
        if response == "yes":
            self.user_consent = True
            print(f"Device {self.device_id}: Consent granted. Data collection enabled.")
            return True
        else:
            self.user_consent = False
            print(f"Device {self.device_id}: Consent denied. Data collection disabled.")
            return False

    def collect_data(self, sensor_name, value):
        """
        Collects data if user consent is granted.

        Args:
            sensor_name (str): Name of the sensor providing the data.
            value: The data value.
        """
        if self.user_consent:
            timestamp = datetime.datetime.now().isoformat()
            if sensor_name not in self.data_collected:
                self.data_collected[sensor_name] = []
            self.data_collected[sensor_name].append({"timestamp": timestamp, "value": value})
            self.last_updated = timestamp
            print(f"Device {self.device_id}: Collected data from '{sensor_name}': {value} at {timestamp}")
        else:
            print(f"Device {self.device_id}: Cannot collect data. User consent not granted.")

    def anonymize_data(self, fields_to_anonymize=None):
        """
        Simulates anonymizing collected data.

        Args:
            fields_to_anonymize (list, optional): List of sensor names to anonymize.
                                                 If None, anonymizes all data.
        """
        if self.user_consent:
            print(f"Device {self.device_id}: Anonymizing collected data...")
            if fields_to_anonymize is None:
                self.data_collected = {"anonymized_data": "Data anonymized"}
            else:
                for field in fields_to_anonymize:
                    if field in self.data_collected:
                        self.data_collected[field] = [{"timestamp": item["timestamp"], "value": "[ANONYMIZED]"} for item in self.data_collected[field]]
            print(f"Device {self.device_id}: Data anonymization complete.")
        else:
            print(f"Device {self.device_id}: No data to anonymize as consent was not granted.")

    def provide_transparency_report(self):
        """Provides a basic report on the device's data collection practices."""
        print(f"\n--- Transparency Report for Device {self.device_id} ({self.device_type}) ---")
        print(f"User Consent Granted: {self.user_consent}")
        if self.user_consent:
            print("Data Collected:")
            if self.data_collected:
                for sensor, readings in self.data_collected.items():
                    print(f"  - {sensor}: {len(readings)} readings")
                    if self.last_updated:
                        print(f"    - Last updated: {self.last_updated}")
            else:
                print("  - No data collected yet.")
        else:
            print("No data has been collected due to lack of consent.")
        print("---------------------------------------------------------\n")

# Example Usage
if __name__ == "__main__":
    thermometer = SmartDevice(device_id="thermometer_123", device_type="temperature_sensor")
    motion_sensor = SmartDevice(device_id="motion_456", device_type="motion_detector")

    thermometer.request_consent()
    motion_sensor.request_consent()

    if thermometer.user_consent:
        thermometer.collect_data("temperature", 25.5)
        time.sleep(2)
        thermometer.collect_data("temperature", 26.0)

    if motion_sensor.user_consent:
        motion_sensor.collect_data("motion", True)
        time.sleep(1)
        motion_sensor.collect_data("motion", False)

    thermometer.provide_transparency_report()
    motion_sensor.provide_transparency_report()

    thermometer.anonymize_data(fields_to_anonymize=["temperature"])
    thermometer.provide_transparency_report()

    motion_sensor.anonymize_data() # Anonymize all data
    motion_sensor.provide_transparency_report()
```
