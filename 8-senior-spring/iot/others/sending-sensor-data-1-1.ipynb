```python
import time
import random
import requests
import json

SENSOR_API_URL = "http://127.0.0.1:5000/sensor_data" # modify it.

def get_sensor_data():
    """Simulates a sensor reading."""
    data = {
        "temperature": round(random.uniform(20, 30), 1),
        "humidity": random.randint(40, 60),
        "timestamp": time.time()
    }
    return data

def send_data(data):
    """Sends sensor data to the API."""
    try:
        response = requests.post(SENSOR_API_URL, json=data)
        response.raise_for_status() # Raise HTTPError for bad responses (4xx or 5xx)
        print(f"Data sent: {data}")
    except requests.exceptions.RequestException as e:
        print(f"Error sending data: {e}")

if __name__ == "__main__":
    while True:
        sensor_data = get_sensor_data()
        send_data(sensor_data)
        time.sleep(2)  # Send data every 2 seconds
```

```python
from flask import Flask, request, jsonify

app = Flask(__name__)

@app.route('/temperature', methods=['POST'])
def receive_temperature():
    """Receives temperature data from the phone."""
    data = request.get_json()
    if data and 'temperature' in data:
        temperature = data['temperature']
        print(f"Received temperature: {temperature} °C")
        # Store or process the temperature data as needed
        return jsonify({"message": "Temperature received"}), 200
    else:
        return jsonify({"error": "Invalid data"}), 400

if __name__ == '__main__':
    app.run(debug=True, host='0.0.0.0') # Make sure your network allows access.
```
