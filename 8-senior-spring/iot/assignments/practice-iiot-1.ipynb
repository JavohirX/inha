```python
import time
import random
import datetime

# Simulate a sensor reading from an industrial machine.  This could be
# temperature, pressure, vibration, etc.  For this example, we'll simulate
# temperature in Celsius.
def get_sensor_data():
    # Simulate some random fluctuation around a normal operating temperature.
    normal_temp = 75  # degrees Celsius
    temperature = normal_temp + random.uniform(-5, 5)
    return temperature

# Process the sensor data.  In a real application, this might involve
# filtering, smoothing, or more complex calculations.  Here, we'll just
# round it to one decimal place for simplicity.
def process_data(data):
    processed_data = round(data, 1)
    return processed_data

# Store the data.  In a real application, this would likely involve
# writing to a database (e.g., InfluxDB, TimescaleDB), a message queue
# (e.g., Kafka, RabbitMQ), or a cloud service.  Here, we'll just
# print it to the console with a timestamp.
def store_data(data):
    timestamp = datetime.datetime.now().isoformat()
    print(f"[{timestamp}] Sensor Data: {data} °C")

# Analyze the data and generate alerts if necessary.  Here, we'll
# define a simple threshold for the temperature and print an alert
# if it's exceeded.
def analyze_data(data):
    temperature_threshold = 85  # degrees Celsius
    if data > temperature_threshold:
        timestamp = datetime.datetime.now().isoformat()
        print(f"[{timestamp}] ALERT! Temperature exceeds threshold: {data} °C > {temperature_threshold} °C")

# Main function to simulate the IoT data pipeline.
def main():
    while True:
        try:
            # 1. Get sensor data
            sensor_data = get_sensor_data()

            # 2. Process the data
            processed_data = process_data(sensor_data)

            # 3. Store the data
            store_data(processed_data)

            # 4. Analyze the data and generate alerts
            analyze_data(processed_data)

            # Wait for a specified interval before the next reading.  In a real
            # application, this interval would depend on the specific requirements
            # of the system (e.g., how frequently data needs to be collected).
            time.sleep(5)  # Read sensor data every 5 seconds
        except Exception as e:
            timestamp = datetime.datetime.now().isoformat()
            print(f"[{timestamp}] Error: {e}")
            time.sleep(5) # Wait and try again.  Important in IOT systems.

if __name__ == "__main__":
    main()
```
