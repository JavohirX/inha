```python
import time
import random

# Simulating an edge device (e.g., sensor)
class EdgeDevice:
    def __init__(self, device_id):
        self.device_id = device_id

    def generate_sensor_data(self):
        # Simulate generating sensor data (e.g., temperature, humidity)
        temperature = random.uniform(20, 30)
        humidity = random.uniform(40, 60)
        return {"temperature": temperature, "humidity": humidity}

    def process_data_locally(self, data):
        # Simulate local data processing (e.g., simple thresholding)
        if data["temperature"] > 28:
            alert = "Temperature high!"
        else:
            alert = "Temperature normal."
        return {"device_id": self.device_id, "processed_data": data, "alert": alert}

# Simulating an edge server
class EdgeServer:
    def aggregate_data(self, processed_data_list):
        # Simulate aggregating data from multiple edge devices
        temperatures = [data["processed_data"]["temperature"] for data in processed_data_list]
        average_temperature = sum(temperatures) / len(temperatures)
        return {"average_temperature": average_temperature}

    def analyze_data(self, aggregated_data):
        # Simulate more complex analysis (e.g., anomaly detection)
        if aggregated_data["average_temperature"] > 27:
            anomaly = "Potential heatwave detected!"
        else:
            anomaly = "No anomalies detected."
        return {"analyzed_data": aggregated_data, "anomaly": anomaly}

# Simulating the cloud (very basic)
class Cloud:
    def store_data(self, final_analysis):
        # Simulate storing data in the cloud
        print(f"Cloud: Storing final analysis: {final_analysis}")

# Main execution
def main():
    devices = [EdgeDevice(device_id=i) for i in range(5)] #5 edge devices.
    edge_server = EdgeServer()
    cloud = Cloud()

    processed_data_list = []

    for device in devices:
        sensor_data = device.generate_sensor_data()
        processed_data = device.process_data_locally(sensor_data)
        processed_data_list.append(processed_data)
        print(f"Device {device.device_id}: {processed_data}")
        time.sleep(0.5) #simulate sensor readings at intervals.

    aggregated_data = edge_server.aggregate_data(processed_data_list)
    analyzed_data = edge_server.analyze_data(aggregated_data)
    print(f"Edge Server: {analyzed_data}")

    cloud.store_data(analyzed_data)

if __name__ == "__main__":
    main()
```
