```python
import numpy as np

def scaled_dot_product_attention(Q, K, V):
    """
    Computes the attention weights and the resulting context vectors.
    """
    # 1. Calculate Dot Product between Q and K^T
    matmul_qk = np.matmul(Q, K.T)

    # 2. Scale by square root of d_k
    d_k = K.shape[-1]
    scaled_attention_logits = matmul_qk / np.sqrt(d_k)

    # 3. Softmax to get probabilities (weights sum to 1)
    # We apply softmax along the last axis
    attention_weights = np.exp(scaled_attention_logits) / np.sum(np.exp(scaled_attention_logits), axis=-1, keepdims=True)

    # 4. Multiply weights by Values
    output = np.matmul(attention_weights, V)

    return output, attention_weights

# Example usage:
# Let's assume we have 3 tokens, each with an embedding size of 4
Q = np.array([[1, 0, 1, 0], [0, 2, 0, 1], [1, 1, 1, 1]])
K = np.array([[1, 0, 1, 0], [0, 2, 0, 1], [1, 1, 1, 1]])
V = np.array([[10, 0, 0, 0], [0, 20, 0, 0], [0, 0, 30, 0]])

output, weights = scaled_dot_product_attention(Q, K, V)

print("Attention Weights:\n", weights)
print("\nOutput (Contextualized Vectors):\n", output)
```
