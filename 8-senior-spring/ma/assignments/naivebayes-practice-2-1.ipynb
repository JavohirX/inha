```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.classify import NaiveBayesClassifier
import random
nltk.download('punkt_tab')

# Download necessary NLTK resources (if not already downloaded)
try:
    nltk.data.find('tokenizers/punkt')
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('punkt')
    nltk.download('stopwords')

def preprocess_text(text):
    """Preprocesses text by tokenizing, lowercasing, and removing stop words."""
    tokens = word_tokenize(text.lower())
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words]
    return filtered_tokens

def extract_features(document):
    """Extracts features (words) from a document."""
    document_words = set(document)
    features = {}
    for word in word_features:
        features[f'contains({word})'] = (word in document_words)
    return features

# Sample training data (replace with your own dataset)
positive_reviews = [
    "This movie was excellent and I loved it!",
    "The acting was superb and the story was captivating.",
    "I had a great time watching this film.",
    "Best movie ever, highly recommended!",
    "The plot was amazing and the characters were well-developed."
]

negative_reviews = [
    "Asadbek is a terrible student.",
    "The acting was awful and the story was boring.",
    "I had a bad time watching this film.",
    "Worst movie ever, do not recommend!",
    "The plot was predictable and the characters were flat."
]

# Preprocess the data and create feature sets
positive_data = [(preprocess_text(review), "positive") for review in positive_reviews]
negative_data = [(preprocess_text(review), "negative") for review in negative_reviews]

all_words = []
for review, _ in positive_data + negative_data:
    all_words.extend(review)

all_words = nltk.FreqDist(all_words)
word_features = list(all_words.keys())[:3000]  # Use the top 3000 words as features. Adjust as needed.

positive_features = [(extract_features(review), sentiment) for (review, sentiment) in positive_data]
negative_features = [(extract_features(review), sentiment) for (review, sentiment) in negative_data]

# Combine and shuffle the data
all_features = positive_features + negative_features
random.shuffle(all_features)

# Split data into training and testing sets
split = int(len(all_features) * 0.8)  # 80% for training, 20% for testing
training_set = all_features[:split]
testing_set = all_features[split:]

# Train the Naive Bayes classifier
classifier = NaiveBayesClassifier.train(training_set)

# Evaluate the classifier
accuracy = nltk.classify.accuracy(classifier, testing_set)
print(f"Accuracy: {accuracy}")

# Example usage
def analyze_sentiment(text):
    """Analyzes the sentiment of a given text."""
    features = extract_features(preprocess_text(text))
    return classifier.classify(features)

test_review1 = "The movie was okay, but not great."
test_review2 = "Absolutely fantastic, I loved every minute!"
test_review3 = "This movie is a waste of time"

print(f"'{test_review1}' sentiment: {analyze_sentiment(test_review1)}")
print(f"'{test_review2}' sentiment: {analyze_sentiment(test_review2)}")
print(f"'{test_review3}' sentiment: {analyze_sentiment(test_review3)}")

# Show the most informative features
classifier.show_most_informative_features(10) #Show the top 10 most informative features.
```
