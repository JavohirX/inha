```python
import numpy as np

def softmax(x):
    """Compute softmax values for each sets of scores in x."""
    e_x = np.exp(x - np.max(x))  # Subtract max for numerical stability
    return e_x / e_x.sum(axis=0)

def rnn_step(input_t, prev_hidden_t, U, W, V):
    """Performs a single RNN step."""
    hidden_t = np.tanh(np.dot(U, input_t) + np.dot(W, prev_hidden_t))
    output_t = softmax(np.dot(V, hidden_t))  # Apply softmax for word probabilities
    return output_t, hidden_t

def rnn_forward(inputs, U, W, V, initial_hidden):
    """Performs the forward pass of the RNN."""
    T = len(inputs)
    hidden_size = W.shape[0]
    output_size = V.shape[0]

    hidden_states = np.zeros((T + 1, hidden_size))
    outputs = np.zeros((T, output_size))

    hidden_states[0] = initial_hidden #use the initial hidden state.

    for t in range(T):
        input_t = inputs[t]
        prev_hidden_t = hidden_states[t]
        outputs[t], hidden_states[t + 1] = rnn_step(input_t, prev_hidden_t, U, W, V)

    return outputs, hidden_states

# Example usage:

# Define vocabulary and word-to-index mappings
vocabulary = ["the", "quick", "brown", "fox", "jumps", "over", "lazy", "dog"]
word_to_index = {word: i for i, word in enumerate(vocabulary)}
index_to_word = {i: word for i, word in enumerate(vocabulary)}
vocab_size = len(vocabulary)

# Example input sequence (one-hot encoded words)
input_sequence = ["the", "quick", "brown", "fox"]
inputs = [np.eye(vocab_size)[word_to_index[word]] for word in input_sequence]

# RNN parameters (randomly initialized)
hidden_size = 5
U = np.random.randn(hidden_size, vocab_size)
W = np.random.randn(hidden_size, hidden_size)
V = np.random.randn(vocab_size, hidden_size)

# Initial hidden state.
initial_hidden = np.zeros((hidden_size,))

# Run the RNN forward pass
outputs, hidden_states = rnn_forward(inputs, U, W, V, initial_hidden)

# Print the predicted words (most likely words)
print("Input sequence:", input_sequence)
predicted_words = [index_to_word[np.argmax(output)] for output in outputs]
print("Predicted words:", predicted_words)

#Example with longer sentence.
long_input = ["the", "quick", "brown", "fox", "jumps", "over", "the"]
long_inputs = [np.eye(vocab_size)[word_to_index[word]] for word in long_input]
long_outputs, long_hidden_states = rnn_forward(long_inputs, U, W, V, initial_hidden)
long_predicted_words = [index_to_word[np.argmax(output)] for output in long_outputs]
print("\nLonger input sequence:", long_input)
print("Longer predicted sequence:", long_predicted_words)
```
