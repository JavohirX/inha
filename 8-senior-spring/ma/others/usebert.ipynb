```python
#pip install transformers torch
from transformers import pipeline

# 1. Initialize the sentiment analysis pipeline
# This automatically loads a BERT-based model (distilbert-base-uncased-finetuned-sst-2-english)
analyzer = pipeline("sentiment-analysis")

# 2. Define your list of sentences
sentences = [
    "I absolutely love how easy it is to use BERT for NLP tasks!",
    "The service at the restaurant was okay, but the food was a bit cold.",
    "I'm so frustrated because the app keeps crashing every five minutes."
]

# 3. Run the analysis
results = analyzer(sentences)

# 4. Display the results
print(f"{'Text':<70} | {'Label':<10} | {'Score'}")
print("-" * 95)

for text, result in zip(sentences, results):
    label = result['label']
    score = result['score']
    print(f"{text:<70} | {label:<10} | {score:.4f}")
```

```python
# Example using a model specifically trained for product reviews
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"
specific_analyzer = pipeline("sentiment-analysis", model=model_name)

result = specific_analyzer("The battery life on this laptop is phenomenal.")
print(result)
# This model outputs star ratings (1-5) instead of POSITIVE/NEGATIVE
```
