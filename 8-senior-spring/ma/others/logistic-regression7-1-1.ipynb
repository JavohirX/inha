```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

# Sample movie reviews
reviews = [
    "This movie was fantastic!",
    "The plot was confusing.",
    "The performances were outstanding.",
    "I didn't enjoy the film at all."
]

# Labels for sample movie reviews
labels = [1, -1, 1, -1]  # Positive (+1) or negative (-1) sentiments

# Convert text data into bag-of-words representation
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(reviews)

# Train logistic regression model
model = LogisticRegression()
model.fit(X_train, labels)

# Predict sentiment of new movie reviews
new_reviews = [
  #  "The movie exceeded my expectations.",
  #  "The screenplay was poorly written.",
    "I did not like this movie",
  "that was a terrible movie"
]
X_test = vectorizer.transform(new_reviews)
predictions = model.predict(X_test)

for review, prediction in zip(new_reviews, predictions):
    sentiment = "positive" if prediction == 1 else "negative"
    print(f"Review: '{review}' -> Predicted sentiment: {sentiment}")
```
