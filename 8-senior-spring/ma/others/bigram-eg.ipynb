```python
import random
from collections import defaultdict

# 1. Prepare and tokenize the corpus
corpus = [
    "Azizbek is a good student",
    "He loves reading book",
    "He does assignment on time",
    "He never miss any class"
]

def build_bigram_model(data):
    # Counts occurrences of word pairs
    model = defaultdict(lambda: defaultdict(int))
    for sentence in data:
        words = ["<START>"] + sentence.split() + ["<END>"]
        for i in range(len(words)-1):
            curr, next_w = words[i], words[i+1]
            model[curr][next_w] += 1
    return model

def generate_text(model, limit=10):
    # Predicts next word based on weighted probability
    sentence = []
    curr = "<START>"

    for _ in range(limit):
        choices = list(model[curr].keys())
        if not choices: break

        # Weighted random choice based on counts
        weights = list(model[curr].values())
        curr = random.choices(choices, weights=weights)[0]

        if curr == "<END>": break
        sentence.append(curr)

    return " ".join(sentence)

# Execution
bigram_model = build_bigram_model(corpus)
print(f"Generated: {generate_text(bigram_model)}")

```

```python
import nltk
nltk.download('punkt_tab')
from nltk.util import bigrams

text = "This is a simple bigram model example"
tokens = nltk.word_tokenize(text.lower())
bi_list = list(bigrams(tokens))

print(bi_list)
# Output: [('this', 'is'), ('is', 'a'), ('a', 'simple'), ...]
```
