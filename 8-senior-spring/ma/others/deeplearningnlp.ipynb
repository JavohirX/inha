```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Flatten, Dense

# Sample data
sentences = ['Tashkent is a great city!',
             'I really enjoyed the sunlight.',
             'The traffic was terrible.',
             'It cost a long time.']
labels = np.array([1, 1, 0, 0])  # 1 for positive sentiment, 0 for negative sentiment

# Tokenize and pad sequences
tokenizer = Tokenizer(num_words=10000, oov_token="<OOV>")
tokenizer.fit_on_texts(sentences)
sequences = tokenizer.texts_to_sequences(sentences)
padded_sequences = pad_sequences(sequences, maxlen=10, padding='post')

# Model architecture
model = Sequential([
    Embedding(input_dim=10000, output_dim=16, input_length=10),
    Flatten(),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# Model training
model.fit(padded_sequences, labels, epochs=10, validation_split=0.2)

# Test with new data
test_sentence = ["I liked the city","due to big park"]
test_sequence = tokenizer.texts_to_sequences(test_sentence)
padded_test_sequence = pad_sequences(test_sequence, maxlen=10, padding='post')
print(model.predict(padded_test_sequence))
print(test_sequence)
```
