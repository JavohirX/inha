```python
import moondream as md
from PIL import Image

# 1. Initialize the local model
# Updated: Pass the model path directly
model = md.vl(model="moondream-latest.mf")

def run_local_vision(image_path):
    try:
        # 2. Load your image
        image = Image.open(image_path)

        # 3. Encode the image (the "vision" part)
        encoded_image = model.encode_image(image)

        # 4. Perform multimodal tasks (Text + Image)
        # Task A: Auto-Captioning
        caption = model.caption(encoded_image)["caption"]
        print(f"Caption: {caption}")

        # Task B: Visual Question Answering (VQA)
        question = "What color is the car in this image?"
        answer = model.query(encoded_image, question)["answer"]
        print(f"Question: {question}")
        print(f"Answer: {answer}")
    except Exception as e:
        print(f"An error occurred: {e}")

# To use: Place an image named 'test.jpg' in your folder and run:
# run_local_vision('test.jpg')
```
