```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer

# Training data
documents = ["great movie", "excellent performance", "terrible plot", "bad acting", "just okay", "the movie was long"]
labels = ["positive", "positive", "negative", "negative", "neutral", "neutral"]

# Feature extraction (convert text to numerical features)
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

# Train the Naive Bayes classifier
classifier = MultinomialNB()
classifier.fit(X, labels)

# New document
new_document = ["the movie was not okay"]
new_X = vectorizer.transform(new_document)

# Predict the sentiment
prediction = classifier.predict(new_X)
print(prediction)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.classify import NaiveBayesClassifier
nltk.download('punkt_tab')

# Download necessary NLTK resources (if not already downloaded)
nltk.download('punkt')
nltk.download('stopwords')

# Sample lexicon (simplified)
lexicon = {
    'positive': ['good', 'great', 'excellent', 'amazing', 'happy', 'love', 'best'],
    'negative': ['bad', 'terrible', 'awful', 'sad', 'hate', 'worst', 'poor'],
    'neutral': ['okay', 'average', 'normal', 'just', 'so-so']
}

# Training data (using lexicon-based features)
training_data = []
for sentiment, words in lexicon.items():
    for word in words:
        training_data.append(({word: True}, sentiment))

# Create Naive Bayes classifier
classifier = NaiveBayesClassifier.train(training_data)

# Function to extract features from a text
def extract_features(text):
    features = {}
    tokens = word_tokenize(text.lower())
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [word for word in tokens if word.isalnum() and word not in stop_words] #remove punctuation and stopwords

    for word in filtered_tokens:
        features[word] = True
    return features

# Example sentences
sentences = [
    "This movie was excellent!",
    "The food was terrible.",
    "It was just okay.",
    "I love this place!",
    "The service was poor.",
    "The weather is normal.",
]

# Perform sentiment analysis
for sentence in sentences:
    features = extract_features(sentence)
    sentiment = classifier.classify(features)
    print(f"Sentence: '{sentence}' - Sentiment: {sentiment}")

# Example of a sentence with words not in the training data
sentence2 = "The performance was outstanding and the plot was engaging."
features2 = extract_features(sentence2)
sentiment2 = classifier.classify(features2)
print(f"Sentence: '{sentence2}' - Sentiment: {sentiment2}")

#Example of a sentence with mixed sentiment.
sentence3 = "The good acting was ruined by the bad plot."
features3 = extract_features(sentence3)
sentiment3 = classifier.classify(features3)
print(f"Sentence: '{sentence3}' - Sentiment: {sentiment3}")
```
