### Kleenee closer
# "*" : 0 or more
# "+" : 1 or more
New (s)* : New, News, Newss, Newssss, Newssssss

New (s)+ : News, Newss, Newssss, Newssssss

```python
import re
pattern="New(s)*"
if re.match(pattern, "New"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="New(s)*"
if re.match(pattern, "Newssssssss"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="New(s)*"
if re.match(pattern, "News"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="New(s)*"
if re.match(pattern, "Ne"):
    print("Match ")
else:
    print("Not Match")
```

## Example: Plus

```python
pattern="New(s)+"
if re.match(pattern, "New"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="New(s)+"
if re.match(pattern, "News"):
    print("Match ")
else:
    print("Not Match")
```

## Meta Character: | (Pipe Symbol)
| : works similiar like logical OR

Go (es|ing): Goes, Going
S (a|u|i) ng: Sing, Sung, Sang

Example: Two string

```python
pattern="Go(es|ing)"
if re.match(pattern, "Goes"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="Go(es|ing)"
if re.match(pattern, "Going"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="Go(es|ing)"
if re.match(pattern, "Go"):
    print("Match ")
else:
    print("Not Match")
```

## Example: Multiple String

```python
pattern="S(a|u|i)ng"
if re.match(pattern, "Sing"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="S(a|u|i)ng"
if re.match(pattern, "Sung"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="S(a|u|i)ng"
if re.match(pattern, "Sang"):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="S(a|u|i)ng"
if re.match(pattern, "Sng"):
    print("Match ")
else:
    print("Not Match")
```

## Example: Multiple String with star

```python
pattern="S(a|u|i)*ng"
if re.match(pattern, "Sng"):
    print("Match ")
else:
    print("Not Match")
```

### Example: Sentence or word

```python
pattern="(Bekzod|Aziz)"
text="Aziz"
if re.match(pattern, text):
    print("Match ")
else:
    print("Not Match")
```

```python
pattern="(Bekzod|Aziz)"
text="Bekzod"
if re.match(pattern, text):
    print("Match ")
else:
    print("Not Match")
```

```python
# Not correct: choose Bekzod or Aziz
pattern="(Bekzod|Aziz)"
text="BekzodAziz"
if re.match(pattern, text):
    print("Match ")
else:
    print("Not Match")
```
