```python
import collections
import re
import random

def mock_llm_generate(prompt):
    """
    Simulates an LLM generating diverse reasoning paths.
    In a real API (OpenAI/Anthropic), you would set temperature > 0.5.
    """
    responses = [
        "The total is 5 + 5 which is 10. Answer: 10",
        "First add 5 and 5 to get 10. Answer: 10",
        "If you have 5 and add 5, you sometimes get 12. Answer: 12", # Hallucination
        "Starting with 5, adding 5 gives 10. Answer: 10",
        "5 plus 5 equals 11. Answer: 11" # Calculation error
    ]
    return random.choice(responses)

def extract_answer(text):
    """
    Extracts the numerical answer following the 'Answer:' label.
    """
    match = re.search(r"Answer:\s*(\d+)", text)
    return match.group(1) if match else None

def self_consistency_logic(question, n_samples=5):
    """
    Implementation of the Self-Consistency decoding strategy.
    """
    results = []

    print(f"Processing Question: {question}")
    print(f"Sampling {n_samples} independent reasoning paths...\n")

    for i in range(n_samples):
        path = mock_llm_generate(question)
        ans = extract_answer(path)
        if ans:
            results.append(ans)
            print(f"Path {i+1}: {path}")

    # Majority Voting
    counts = collections.Counter(results)
    final_answer, votes = counts.most_common(1)[0]

    return final_answer, votes

# --- Execute ---
question = "What is 5 + 5?"
final_ans, vote_count = self_consistency_logic(question)

print("-" * 30)
print(f"Majority Consensus Answer: {final_ans}")
print(f"Confidence: {vote_count} out of 5 samples agreed.")
```
