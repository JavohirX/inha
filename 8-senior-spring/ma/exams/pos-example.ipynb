```python
import numpy as np

def binary_cross_entropy(y_true, y_pred):
    """
    Calculates binary cross-entropy loss.

    Args:
        y_true: True labels (0 or 1).
        y_pred: Predicted probabilities (between 0 and 1).

    Returns:
        The binary cross-entropy loss.
    """
    epsilon = 1e-15  # Small value to avoid log(0)
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)  # Clip to prevent log(0)
    loss = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return loss

# Example usage
y_true = np.array([1, 0, 1, 0])  # True labels
y_pred1 = np.array([0.9, 0.1, 0.8, 0.2])  # Good predictions
y_pred2 = np.array([0.3, 0.7, 0.1, 0.9])  # Poor predictions

loss1 = binary_cross_entropy(y_true, y_pred1)
loss2 = binary_cross_entropy(y_true, y_pred2)

print(f"Loss for good predictions: {loss1}")
print(f"Loss for poor predictions: {loss2}")

# Example of a very confident wrong prediction.
y_pred3 = np.array([0.1, 0.1, 0.1, 0.1]) #all wrong predictions.
loss3 = binary_cross_entropy(y_true, y_pred3)
print(f"Loss for very confident wrong predictions: {loss3}")
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

def simple_pos_tagging(text):
    """
    Performs simple POS tagging using NLTK.

    Args:
        text: The input text string.

    Returns:
        A list of (word, tag) tuples.
    """
    tokens = nltk.word_tokenize(text)
    tagged_words = nltk.pos_tag(tokens)
    return tagged_words

# Example usage
text = "The quick brown fox jumps over the lazy dog."
tagged_text = simple_pos_tagging(text)

print(tagged_text)

text2 = "I saw a saw."
tagged_text2 = simple_pos_tagging(text2)
print(tagged_text2)
```

```python
import spacy

# Load the simple English language model
nlp = spacy.load("en_core_web_sm")

def process_text(text):
    """
    Processes text using SpaCy's simple model.

    Args:
        text: The input text string.

    Returns:
        A SpaCy Doc object.
    """
    doc = nlp(text)
    return doc

# Example usage
text = "SpaCy is a powerful NLP library. It's easy to use."
doc = process_text(text)

# Access tokens and their attributes
for token in doc:
    print(f"Token: {token.text}, POS: {token.pos_}, Tag: {token.tag_}, Lemma: {token.lemma_}")

# Access sentences
for sent in doc.sents:
    print(f"Sentence: {sent.text}")

# Access entities
for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```
