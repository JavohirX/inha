## **Activation Functions**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### Sigmoid function
def sigmoid(z):
    return 1/(1+np.exp(-z))

### Tanh function
def tanh(z):
    pos_exp = np.exp(z)
    neg_exp = np.exp(-z)
    return (pos_exp - neg_exp)/(pos_exp + neg_exp)

### Rectified Linear Unit (ReLU) Function
def relu(z):
    return np.maximum(0, z)

### Leaky ReLU
def leaky_relu(z, alpha=0.01):
    return np.maximum(z, alpha*z)

### Swish function
def swish(z):
    return z/(1+np.exp(-z))

### Exponential Linear Unit (ELU) function
def elu(z, alpha=1.0):
    return np.where(z > 0, z, alpha * (np.exp(z) - 1))

### Scaled Exponential Linear Unit (SELU) 
def selu(z):
    alpha = 1.6732632423543772848170429916717
    scale = 1.0507009873554804934193349852946

    # Use np.where to apply different formulas based on the sign of x
    return scale * np.where(z > 0, z, alpha * (np.exp(z) - 1))
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
plt.plot(x, y)
plt.xlabel("Values of x")
plt.ylabel("Values of y")
plt.title("Original data")
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
plt.plot(x, sigmoid(y))
plt.xlabel("Values of x")
plt.ylabel("Values of sigmoid function")
plt.title("Sigmoid function output")
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
plt.plot(x, tanh(y))
plt.xlabel("Values of x")
plt.ylabel("Values of Tanh function")
plt.title("Tanh function output")
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
plt.plot(x, relu(y))
plt.xlabel("Values of x")
plt.ylabel("Values of ReLU function")
plt.title("ReLU function output")
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
plt.plot(x, elu(y, alpha=0.1))
plt.xlabel("Values of x")
plt.ylabel("Values of ELU function")
plt.title("ELU function output")
plt.grid()
plt.show()
```

```python
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
plt.plot(x, selu(y))
plt.xlabel("Values of x")
plt.ylabel("Values of SELU function")
plt.title("SELU function output")
plt.grid()
plt.show()
```

```python
def softmax(z):
    return np.exp(z)/np.sum(np.exp(z))
```

```python
input_logits = np.array([2.0, 4.0, 1.0, 0.5])
output_probabilities = softmax(input_logits)

print("Input Logits:", input_logits)
print("Softmax Probabilities:", output_probabilities)
print("Sum of Probabilities:", np.sum(output_probabilities))
```
